# Library Book Borrowing System
This notebook contains the corrected solution for the console-based Library Book Borrowing System.

```python
from datetime import datetime
```

```python
from datetime import datetime

class LibraryManager:
    def __init__(self):
        self.books = {}  # Available books: book_id -> title
        self.borrowed = {}  # Borrowed books: book_id -> (borrower_name, borrow_date)
        self.file_name = "BorrowedBooks.txt"
        self.backup_template = "BorrowedBooks_Backup_{}.txt"

    def add_book(self, book_id, title):
        if book_id in self.books:
            print("Book ID already exists.")
        else:
            self.books[book_id] = title
            print(f"Book '{title}' added.")

    def remove_book(self, book_id):
        if book_id in self.books:
            removed = self.books.pop(book_id)
            print(f"Book '{removed}' removed.")
        else:
            print("Book ID not found.")

    def display_books(self):
        if not self.books:
            print("No books available.")
        else:
            print("Available books:")
            for book_id, title in self.books.items():
                print(f"{book_id}: {title}")

    def borrow_book(self, book_id, borrower):
        if book_id in self.books and book_id not in self.borrowed:
            self.borrowed[book_id] = (borrower, datetime.now())
            print(f"'{self.books[book_id]}' borrowed by {borrower}.")
        else:
            print("Book not available or already borrowed.")

    def view_borrowed_books(self):
        if not self.borrowed:
            print("No books are currently borrowed.")
        else:
            print("Borrowed books:")
            for book_id, (borrower, date) in self.borrowed.items():
                print(f"{book_id}: {self.books.get(book_id, 'Unknown Title')} borrowed by {borrower} on {date.strftime('%Y-%m-%d')}")

    def return_book(self, book_id):
        if book_id in self.borrowed:
            borrower, borrow_date = self.borrowed.pop(book_id)
            days = (datetime.now() - borrow_date).days
            fine = max(0, (days - 7) * 2)  # $2 per late day after 7 days
            print(f"{borrower} returned '{self.books.get(book_id, 'Unknown Title')}'. Fine: ${fine}")
        else:
            print("Book is not currently borrowed.")

    def save_borrowed_to_file(self):
        try:
            with open(self.file_name, "w") as f:
                for book_id, (borrower, date) in self.borrowed.items():
                    line = f"{book_id},{borrower},{date.strftime('%Y-%m-%d')}\n"
                    f.write(line)
            print("Borrowed book list saved.")
        except Exception as e:
            print(f"Error saving to file: {e}")

    def display_file(self):
        try:
            with open(self.file_name, "r") as f:
                content = f.read()
                print("Borrowed Book File Contents:")
                print(content)
        except FileNotFoundError:
            print("File not found.")

    def backup_and_clear_file(self):
        try:
            with open(self.file_name, "r") as f:
                content = f.read()
            timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
            backup_file = self.backup_template.format(timestamp)
            with open(backup_file, "w") as bf:
                bf.write(content)
            open(self.file_name, "w").close()
            print(f"Backup created at '{backup_file}' and file cleared.")
        except FileNotFoundError:
            print("File not found to backup.")
        except Exception as e:
            print(f"Error during backup: {e}")
```

```python

def menu():
    lib = LibraryManager()
    while True:
        print("\n=== Library Menu ===")
        print("1. Add Book")
        print("2. Remove Book")
        print("3. Display Available Books")
        print("4. Borrow Book")
        print("5. View Borrowed Books")
        print("6. Return Book")
        print("7. Save Borrowed List to File")
        print("8. Display File Content")
        print("9. Backup and Clear File")
        print("0. Exit")
        print("=====================")
        
        try:
            choice = input("Enter choice (0-9): ")
            if choice == "1":
                book_id = input("Enter book ID: ")
                title = input("Enter book title: ")
                lib.add_book(book_id, title)
            elif choice == "2":
                book_id = input("Enter book ID to remove: ")
                lib.remove_book(book_id)
            elif choice == "3":
                lib.display_books()
            elif choice == "4":
                book_id = input("Enter book ID to borrow: ")
                borrower = input("Enter borrower name: ")
                lib.borrow_book(book_id, borrower)
            elif choice == "5":
                lib.view_borrowed_books()
            elif choice == "6":
                book_id = input("Enter book ID to return: ")
                lib.return_book(book_id)
            elif choice == "7":
                lib.save_borrowed_to_file()
            elif choice == "8":
                lib.display_file()
            elif choice == "9":
                lib.backup_and_clear_file()
            elif choice == "0":
                print("Goodbye!")
                break
            else:
                print("Invalid input. Please enter a number 0-9.")
        except Exception as e:
            print(f"An error occurred: {e}")
```

```python
menu()
```
